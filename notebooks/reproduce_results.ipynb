# Reproduce: Coordination-Adjusted Architecture Selection

Synthetic scenarios; coefficients are hypotheses.

```python
from pathlib import Path
import json,sys
import pandas as pd
repo=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd()
sys.path.insert(0,str(repo/'src'))
from agent_arch_select import recommend_architecture
rows=[json.loads(x) for x in (repo/'data/architecture_scenarios_v0_2.jsonl').read_text().splitlines()]
df=pd.DataFrame(rows)
features=['decomposability','specialization_need','verification_value','coordination_burden','error_propagation_risk','resource_sensitivity']
df['recommended']=df.apply(lambda r: recommend_architecture({c:r[c] for c in features}),axis=1)
df[['scenario_id','recommended']].head()
```
